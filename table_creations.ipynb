```python
import pyspark.sql

spark.sql(
    """
    CREATE TABLE IF NOT EXISTS organization.employee_events (
        emp_id INT,
        name STRING,
        dept STRING,
        salary DOUBLE,
        event_type STRING,
        even_ts TIMESTAMP,
        source STRING
    )
    USING DELTA;
    """
)

spark.sql(
    """
    CREATE TABLE IF NOT EXISTS organization.department(
        dept STRING,
        location STRING,
        effective_from DATE
    )
    USING DELTA;
    """
)

spark.sql(
    """
    CREATE TABLE IF NOT EXISTS organization.pipeline_runs (
        run_id INT,
        run_ts TIMESTAMP
    )
    USING DELTA;
    """
)


```
